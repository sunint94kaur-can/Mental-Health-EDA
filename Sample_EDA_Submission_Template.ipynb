## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.



#Steps to make dataset ready for analysis. 1. Convert datatypes
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
# 2. Handle outliers in age
df = df[(df['Age'] >= 18) & (df['Age'] <= 100)]

#3. Standardize Gender Categories
df['Gender'] = df['Gender'].str.strip().str.lower()

male_vals = ['male','m','male-ish', 'maile', 'cis male','mal','male(cis)']
female_vals = ['female','f','cis female','woman','femake']
df['Gender'] = df['Gender'].apply(
    lambda x : 'Male' if x in male_vals else ('Female' if x in female_vals else 'Other')
)

#4. Handle Missing Values
df['state'] = df['state'].fillna('Not Applicable')
df['self_employed'] = df['self_employed'].fillna('Unknown')
df['work_interfere'] = df['work_interfere'].fillna('Not Applicable')

#5. Reset index after filtering rows
df = df.reset_index(drop = True)

#6. Verify the result
print(df.shape)
print(df.dtypes)
print(df.isna().sum())
````

#### Chart - 14 - Correlation Heatmap

````python
# Correlation Heatmap visualization code

df['treatment_map'] = df['treatment'].map({'Yes': 1, 'No': 0})
df['tech_map'] = df['tech_company'].map({'Yes': 1, 'No': 0})
df['remote_map'] = df['remote_work'].map({'Yes': 1, 'No': 0})
df['family_history_map'] = df['family_history'].map({'Yes': 1, 'No': 0})

corr_cols = ['Age','treatment_map','tech_map','remote_map','family_history_map']
print(corr_cols)
plt.figure(figsize=(6,5))
sns.heatmap(df[corr_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
````

All other correlations are weak (close to 0), including Age, tech_company, and remote_work — none of them show a meaningful relationship with treatment-seeking or with each other.
